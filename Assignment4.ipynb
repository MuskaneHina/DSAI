```python
# Q1
users = ["Ali", "Sara", "John", "Zara", "Mike"]

count = 0

for user in users:
    if count == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
    count += 1


# Q2
temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
    print(f"Temperature: {temp}")


# Q3
cart = [100, 250, 50, 300]
total = 0

for price in cart:
    total += price

print(f"Total Price: {total}")


# Q4
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue
    print(f"Access granted to: {customer['name']}")


# Q5
users = ["Ali", "Sara", "John"]

for user in users:
    print(f"Hello {user}, your subscription is active!")


# Q6
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []

for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print("domains:", domains)


# Q7
numbers = [10, 12, 15, 17, 20, 22, 25, 30, 31, 33]

result = [num for num in numbers if num % 3 == 0 or num % 5 == 0]

print("Numbers divisible by 3 or 5:", result)


# Q8
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for num in numbers:
    for multiplier in multipliers:
        print(f"{num} x {multiplier} = {num * multiplier}")


# Q9
correct_password = "python123"
password = ""

while password != correct_password:
    password = input("Enter password: ")

    if password == correct_password:
        print("Access granted!")
    else:
        print("Incorrect password. Try again.")


# Q10
countdown = 5

while countdown >= 1:
    print(countdown)
    countdown -= 1

print("Time's up!")
```
